```python
import sys, json
sys.path.insert(0, '..')
import pandas as pd
import matplotlib.pyplot as plt

from src.utils.config import load_settings
from src.data.load_dataset import load_raw

settings = load_settings()
df = load_raw(settings.data_path, nrows=50000)
df.head()
```
